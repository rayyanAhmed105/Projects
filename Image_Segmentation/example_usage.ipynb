## 3. Configure Training Parameters

```python
# Model configuration
ENCODER = 'resnet34'  # Can be: resnet18, resnet34, resnet50, efficientnet-b0, etc.
ENCODER_WEIGHTS = 'imagenet'  # Pretrained weights
ARCHITECTURE = 'Unet'  # Can be: Unet, UnetPlusPlus, FPN, PSPNet, DeepLabV3, DeepLabV3Plus, etc.
ACTIVATION = 'sigmoid'  # 'sigmoid' for binary, 'softmax' for multiclass

# Training configuration
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
BATCH_SIZE = 16
NUM_EPOCHS = 40
LEARNING_RATE = 0.0001

# Image configuration
IMAGE_HEIGHT = 320
IMAGE_WIDTH = 320
NUM_CLASSES = 1  # 1 for binary segmentation, >1 for multiclass

# Checkpoint directory
CHECKPOINT_DIR = './checkpoints'
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

print(f"Device: {DEVICE}")
print(f"Encoder: {ENCODER}")
print(f"Architecture: {ARCHITECTURE}")
print(f"Image size: {IMAGE_HEIGHT}x{IMAGE_WIDTH}")
print(f"Number of classes: {NUM_CLASSES}")
```

## 4. Create Model

```python
# Create segmentation model
model = smp.Unet(
    encoder_name=ENCODER,
    encoder_weights=ENCODER_WEIGHTS,
    classes=NUM_CLASSES,
    activation=ACTIVATION,
)

print(f"Model created: {ARCHITECTURE} with {ENCODER} encoder")
print(f"Total parameters: {sum(p.numel() for p in model.parameters()):,}")
```
